```python
import pandas as pd

df = pd.read_csv('student-record.csv')

# Clean: Convert column to numeric (handling any noise/spaces)
df['Total_Marks'] = pd.to_numeric(df['Total_Marks'], errors='coerce')

# Normalize: Apply Min-Max formula to scale values between 0 and 1
min_val = df['Total_Marks'].min()
max_val = df['Total_Marks'].max()

df['Total_Marks_Normalized'] = (df['Total_Marks'] - min_val) / (max_val - min_val)

print(df[['Total_Marks', 'Total_Marks_Normalized']].head(10))
```
